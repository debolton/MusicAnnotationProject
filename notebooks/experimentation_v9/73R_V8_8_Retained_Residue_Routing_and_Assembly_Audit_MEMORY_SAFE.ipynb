# Music Annotation Project
## Notebook 73R: V8.8 Retained-Residue Routing and Assembly Audit

**Runtime: CPU. Read-only diagnostic.** Run after the completed Notebook 72 prediction lock.

This notebook replaces the failed treatment-strength counterfactual. It does **not** tune thresholds, change SAFE_REMOVE, generate a new correction policy or rerun learned inference.

### Why this audit exists
The historical V8.8 operational contract is:

```text
M5_REMOVE = M3 & ~SAFE_REMOVE
reconstructed = reconstruct_residual(observed, M5_REMOVE)
out = reconstructed
out[SAFE_REMOVE] = paper[SAFE_REMOVE]
```

So SAFE_REMOVE is a set of pixels that receive direct paper replacement and are removed from the residual inpainting mask. Pixels that are **not** SAFE_REMOVE remain in `M5_REMOVE` and are reconstructed through the residual-mask path.

The current unified `v8_8_operator.py` must therefore be audited against that historical assembly contract before any new cleanup experiment is designed.

### Questions answered
For all 151 Notebook 72 pages:

1. Does the saved Notebook 72 `corrected.png` equal the current operator's assembly semantics?
2. What would the historical Notebook 53 M5_REMOVE assembly produce from the *same frozen* M3 and SAFE_REMOVE maps?
3. Where does visible retained darkness sit: SAFE_REMOVE, M5_REMOVE/fallback, V7.8 preserve, strict PRINT protection, V7.8 remove, or V7.8 ambiguous?
4. How much retained darkness is actually attributable to assembly semantics rather than model selection?
5. Which pages have the largest difference between the current output and the historical M5_REMOVE assembly?

No clean page or truth-derived data is read.

## 2. Integrity helpers and closed-lock verification

````python
def require(x,msg):
    if not bool(x): raise RuntimeError(msg)
def sha256_file(path):
    h=hashlib.sha256()
    with open(path,'rb') as f:
        for b in iter(lambda:f.read(1048576),b''): h.update(b)
    return h.hexdigest()
def array_hash(a):
    a=np.ascontiguousarray(a); h=hashlib.sha256(); h.update(str(a.dtype).encode()); h.update(str(a.shape).encode()); h.update(a.tobytes()); return h.hexdigest()
def atomic_json(path,x):
    t=Path(str(path)+'.tmp'); t.write_text(json.dumps(x,indent=2,sort_keys=True,default=str)); t.replace(path)
def read_gray(path):
    x=cv2.imread(str(path),cv2.IMREAD_GRAYSCALE); require(x is not None,f'Unreadable image: {path}'); return x

def padded_components(mask,margin=CROP_MARGIN,min_area=8):
    n,lab,stats,_=cv2.connectedComponentsWithStats(mask.astype(np.uint8),8); h,w=mask.shape; boxes=[]
    for i in range(1,n):
        x,y,bw,bh,a=stats[i]
        if a>=min_area: boxes.append((max(0,x-margin),max(0,y-margin),min(w,x+bw+margin),min(h,y+bh+margin)))
    changed=True
    while changed:
        changed=False; merged=[]
        while boxes:
            ax0,ay0,ax1,ay1=boxes.pop(); rest=[]
            for bx0,by0,bx1,by1 in boxes:
                if not (ax1<bx0 or bx1<ax0 or ay1<by0 or by1<ay0):
                    ax0=min(ax0,bx0); ay0=min(ay0,by0); ax1=max(ax1,bx1); ay1=max(ay1,by1); changed=True
                else: rest.append((bx0,by0,bx1,by1))
            boxes=rest; merged.append((ax0,ay0,ax1,ay1))
        boxes=merged
    return boxes

def reconstruct_residual(observed,mask):
    result=observed.copy()
    for x0,y0,x1,y1 in padded_components(mask):
        im=observed[y0:y1,x0:x1]; mk=(mask[y0:y1,x0:x1].astype(np.uint8)*255)
        cand=cv2.inpaint(im,mk,3,cv2.INPAINT_TELEA); local=result[y0:y1,x0:x1]; local[mk>0]=cand[mk>0]; result[y0:y1,x0:x1]=local
    out=observed.copy(); out[mask]=result[mask]
    require(np.array_equal(out[~mask],observed[~mask]),'Residual reconstruction changed outside mask.')
    return out

lock=json.loads((SOURCE/'prediction_lock.json').read_text()); verdict=json.loads((SOURCE/'notebook72_verdict.json').read_text())
require(lock.get('state')=='CLOSED','Notebook 72 prediction lock is not closed.')
require(verdict.get('first_failed_gate') is None,'Notebook 72 has a failed gate.')
roster=pd.read_csv(SOURCE/'inputs/runtime_roster.csv',dtype=str).fillna(''); hashes=pd.read_csv(SOURCE/'tables/page_output_hashes.csv',dtype=str).fillna('').set_index('page_key')
require(lock['payload']['runtime_roster_sha256']==sha256_file(SOURCE/'inputs/runtime_roster.csv'),'Roster changed after lock.')
require(lock['payload']['page_output_hashes_sha256']==sha256_file(SOURCE/'tables/page_output_hashes.csv'),'Output-hash table changed after lock.')
require(len(roster)==len(hashes),'Roster/hash count mismatch.')
source_identity={'created_utc':datetime.now(timezone.utc).isoformat(),'source_run':str(SOURCE),'prediction_lock_id':lock['lock_id'],'operator_bundle_id':lock['payload']['operator_bundle_id'],'pages':len(roster),'clean_truth_read':False}
atomic_json(INPUTS/'source_identity.json',source_identity); print(json.dumps(source_identity,indent=2))
````

## 3. Reconstruct decision/routing maps from locked Notebook 72 artefacts

````python
# MEMORY-SAFE: validate one page at a time; never retain full-page arrays across the roster.
integrity=[]
roster_lookup={f'{r.population}__{r.page}':r for _,r in roster.iterrows()}

def locked_paths(key):
    return SOURCE/'pages'/key

def load_locked(key, need_features=False):
    d=locked_paths(key)
    z={
        'observed':read_gray(d/'observed.png'),
        'current':read_gray(d/'corrected.png'),
        'baseline':read_gray(d/'baseline_v78.png'),
        'prepared':read_gray(d/'prepared_mask.png')>0,
        'safe':read_gray(d/'safe_remove.png')>0,
        'm3':read_gray(d/'m3.png')>0,
        'm5':read_gray(d/'m5_remove.png')>0,
        'reason':read_gray(d/'reason_map.png'),
    }
    if need_features:
        z['features']=pd.read_parquet(d/'observable_features.parquet')
    return z

for key,r in roster_lookup.items():
    z=load_locked(key,need_features=True)
    observed=z['observed']; current=z['current']; baseline=z['baseline']; prepared=z['prepared']; safe=z['safe']; m3=z['m3']; m5_saved=z['m5']; reason=z['reason']; features=z['features']
    h=hashes.loc[key]
    require(array_hash(observed)==h.observed_array_sha256,f'{key}: observed hash mismatch')
    require(array_hash(current)==h.corrected_array_sha256,f'{key}: current corrected hash mismatch')
    require(array_hash(baseline)==h.baseline_array_sha256,f'{key}: baseline hash mismatch')
    require(array_hash(safe)==h.safe_remove_array_sha256,f'{key}: safe hash mismatch')
    require(array_hash(m3)==h.m3_array_sha256,f'{key}: m3 hash mismatch')
    require(array_hash(m5_saved)==h.m5_remove_array_sha256,f'{key}: m5 hash mismatch')
    require(not np.any(safe&~m3),f'{key}: SAFE_REMOVE outside M3')
    require(np.array_equal(m3&~safe,m5_saved),f'{key}: saved M5_REMOVE != M3 & ~SAFE_REMOVE')
    preserve=reason==1; strict_print=reason==2; fallback=reason==3; safe_reason=reason==4
    require(np.array_equal(safe,safe_reason),f'{key}: reason SAFE_REMOVE mismatch')
    require(np.array_equal(m5_saved,fallback),f'{key}: reason fallback/M5 mismatch')
    integrity.append({'page_key':key,'population':str(r.population),'m5_equals_m3_minus_safe':True,'safe_subset_m3':True,'outside_current_drift':int(np.sum(current[~prepared]!=observed[~prepared])),'features_rows':len(features)})
    del z,observed,current,baseline,prepared,safe,m3,m5_saved,reason,features,preserve,strict_print,fallback,safe_reason

pd.DataFrame(integrity).to_csv(AUDIT/'locked_state_integrity.csv',index=False)
import gc; gc.collect()
print('Verified locked pages:',len(integrity))
````

## 4. Historical Notebook 53 M5_REMOVE assembly on the exact frozen decisions

````python
# MEMORY-SAFE: historical assembly is generated page-by-page and only scalar summaries survive.
assembly_rows=[]
for key,r in roster_lookup.items():
    z=load_locked(key,need_features=True)
    observed=z['observed']; current=z['current']; m3=z['m3']; safe=z['safe']; m5=z['m5']; features=z['features']
    paper=np.full(observed.shape,np.nan,np.float32)
    yy=features.y.astype(int).to_numpy(); xx=features.x.astype(int).to_numpy()
    paper[yy,xx]=pd.to_numeric(features.paper_value,errors='raise').to_numpy(np.float32)
    require(np.isfinite(paper[safe]).all(),f'{key}: SAFE_REMOVE missing paper values')
    reconstructed=reconstruct_residual(observed,m5)
    historical=reconstructed.copy(); historical[safe]=np.clip(np.rint(paper[safe]),0,255).astype(np.uint8)
    require(np.array_equal(historical[~m3],observed[~m3]),f'{key}: historical assembly changed outside M3')
    diff=current!=historical
    d=PAGES/key; d.mkdir(parents=True,exist_ok=False)
    cv2.imwrite(str(d/'current_notebook72.png'),current)
    cv2.imwrite(str(d/'historical_m5_assembly.png'),historical)
    cv2.imwrite(str(d/'difference_current_vs_historical.png'),diff.astype(np.uint8)*255)
    cv2.imwrite(str(d/'M3.png'),m3.astype(np.uint8)*255)
    cv2.imwrite(str(d/'SAFE_REMOVE.png'),safe.astype(np.uint8)*255)
    cv2.imwrite(str(d/'M5_REMOVE.png'),m5.astype(np.uint8)*255)
    assembly_rows.append({'page_key':key,'population':str(r.population),'m3_pixels':int(m3.sum()),'safe_remove_pixels':int(safe.sum()),'m5_remove_pixels':int(m5.sum()),'current_vs_historical_different_pixels':int(diff.sum()),'different_inside_safe_remove':int(np.sum(diff&safe)),'different_inside_m5_remove':int(np.sum(diff&m5)),'current_equals_historical':bool(not diff.any()),'current_array_sha256':array_hash(current),'historical_array_sha256':array_hash(historical)})
    del z,observed,current,m3,safe,m5,features,paper,reconstructed,historical,diff,yy,xx
    gc.collect()
assembly=pd.DataFrame(assembly_rows); assembly.to_csv(TABLES/'current_vs_historical_m5_assembly.csv',index=False)
print(assembly.groupby('population').agg(pages=('page_key','size'),different_pixels=('current_vs_historical_different_pixels','sum'),pages_different=('current_equals_historical',lambda x:int((~x).sum()))))
````

## 5. Partition retained darkness by exact runtime disposition and V7.8 route evidence

````python
# MEMORY-SAFE: write retained-pixel detail per page instead of accumulating millions of Python dicts.
PIXEL_PARTS=TABLES/'retained_pixel_parts'; PIXEL_PARTS.mkdir(exist_ok=False)
page_rows=[]; summary_rows=[]; retained_pixel_count=0
for key,r in roster_lookup.items():
    z=load_locked(key,need_features=True)
    obs=z['observed']; current=z['current']; baseline=z['baseline']; safe=z['safe']; m3=z['m3']; m5=z['m5']; reason=z['reason']; f=z['features']
    preserve=reason==1; strict=reason==2
    paper=np.full(obs.shape,np.nan,np.float32); p_remove=np.full(obs.shape,np.nan,np.float32); p_print=np.full(obs.shape,np.nan,np.float32)
    yy=f.y.astype(int).to_numpy(); xx=f.x.astype(int).to_numpy()
    paper[yy,xx]=pd.to_numeric(f.paper_value,errors='raise').to_numpy(np.float32)
    p_remove[yy,xx]=pd.to_numeric(f.p_remove,errors='coerce').to_numpy(np.float32)
    p_print[yy,xx]=pd.to_numeric(f.p_print,errors='coerce').to_numpy(np.float32)
    lp=pd.to_numeric(f.local_preserve_fraction,errors='coerce').fillna(0).to_numpy(); lr=pd.to_numeric(f.local_remove_fraction,errors='coerce').fillna(0).to_numpy(); la=pd.to_numeric(f.local_ambiguous_fraction,errors='coerce').fillna(0).to_numpy()
    dominant=np.where((lr>=lp)&(lr>=la),'V78_REMOVE_CONTEXT',np.where((la>=lp)&(la>=lr),'V78_AMBIGUOUS_CONTEXT','V78_PRESERVE_CONTEXT'))
    route_lookup={(int(y),int(x)):str(v) for y,x,v in zip(yy,xx,dominant)}
    residual=np.maximum(paper-current.astype(np.float32),0); valid=m3&np.isfinite(paper)
    disposition=np.full(obs.shape,'OUTSIDE_M3',object); disposition[m5]='M5_REMOVE_RECONSTRUCT'; disposition[safe]='SAFE_REMOVE_PAPER'; disposition[preserve]='V78_PRESERVE'; disposition[strict]='STRICT_PRINT_PROTECT'
    for name,mask in [('SAFE_REMOVE_PAPER',safe),('M5_REMOVE_RECONSTRUCT',m5),('V78_PRESERVE',preserve),('STRICT_PRINT_PROTECT',strict)]:
        q=mask&np.isfinite(paper)
        if not q.any(): continue
        vals=residual[q]
        page_rows.append({'page_key':key,'population':str(r.population),'disposition':name,'pixels_with_paper':int(q.sum()),'residual_10plus':int((vals>=10).sum()),'residual_20plus':int((vals>=20).sum()),'residual_40plus':int((vals>=40).sum()),'mean_residual':float(vals.mean()),'p90_residual':float(np.percentile(vals,90)),'residual_mass':float(vals.sum())})
    ry,rx=np.where(valid&(residual>=10)); retained_pixel_count += len(ry)
    if len(ry):
        part=pd.DataFrame({'page_key':key,'population':str(r.population),'y':ry.astype(np.int32),'x':rx.astype(np.int32),'observed':obs[ry,rx].astype(np.uint8),'current':current[ry,rx].astype(np.uint8),'baseline_v78':baseline[ry,rx].astype(np.uint8),'paper_value':paper[ry,rx].astype(np.float32),'residual_darkness':residual[ry,rx].astype(np.float32),'disposition':disposition[ry,rx].astype(str),'p_remove':p_remove[ry,rx].astype(np.float32),'p_print':p_print[ry,rx].astype(np.float32),'safe_remove':safe[ry,rx],'m5_remove':m5[ry,rx]})
        part['route_context']=[route_lookup.get((int(y),int(x)),'UNKNOWN') for y,x in zip(ry,rx)]
        part.to_parquet(PIXEL_PARTS/f'{key}.parquet',index=False)
        q=part.groupby(['disposition','route_context'],as_index=False).agg(pixels=('residual_darkness','size'),pages=('page_key','nunique'),source_groups=('population','nunique'),mean_residual=('residual_darkness','mean'),residual_mass=('residual_darkness','sum'),mean_p_remove=('p_remove','mean'),mean_p_print=('p_print','mean'))
        # p90 is page-partition local; global p90 is intentionally omitted in memory-safe aggregate.
        summary_rows.append(q)
        del part,q
    del z,obs,current,baseline,safe,m3,m5,reason,f,preserve,strict,paper,p_remove,p_print,yy,xx,lp,lr,la,dominant,route_lookup,residual,valid,disposition,ry,rx
    gc.collect()

disposition_pages=pd.DataFrame(page_rows); disposition_pages.to_csv(TABLES/'residual_by_disposition_and_page.csv',index=False)
if summary_rows:
    ss=pd.concat(summary_rows,ignore_index=True)
    summary=ss.groupby(['disposition','route_context'],as_index=False).agg(pixels=('pixels','sum'),pages=('pages','sum'),source_groups=('source_groups','max'),mean_residual=('mean_residual',lambda x:np.nan),residual_mass=('residual_mass','sum'))
    # exact weighted means from page summaries
    for i,row in summary.iterrows():
        parts=ss[(ss.disposition==row.disposition)&(ss.route_context==row.route_context)]
        w=parts.pixels.to_numpy(float)
        summary.at[i,'mean_residual']=float(np.average(parts.mean_residual,weights=w))
    summary=summary.sort_values('residual_mass',ascending=False)
else:
    summary=pd.DataFrame(columns=['disposition','route_context','pixels','pages','source_groups','mean_residual','residual_mass'])
summary.to_csv(TABLES/'retained_residue_reason_summary.csv',index=False)
atomic_json(TABLES/'retained_pixel_dataset.json',{'format':'partitioned_parquet','directory':str(PIXEL_PARTS),'parts':len(list(PIXEL_PARTS.glob('*.parquet'))),'rows':int(retained_pixel_count),'reason':'memory-safe replacement for one monolithic retained-pixel dataframe'})
print(summary.to_string(index=False)); print('Retained detail rows written:',retained_pixel_count)
````

## 6. Diagnostic overlays and highest-opportunity pages

````python
def overlay(gray,layers,alpha=.58):
    base=cv2.cvtColor(gray,cv2.COLOR_GRAY2BGR); out=base.copy()
    for mask,bgr in layers:
        colour=np.zeros_like(base); colour[:]=bgr; out[mask]=cv2.addWeighted(base,1-alpha,colour,alpha,0)[mask]
    return out

# MEMORY-SAFE: regenerate only one page at a time. Red=M5, green=SAFE_REMOVE, blue=strict PRINT, orange=V7.8 preserve.
page_scores=[]
for key,r in roster_lookup.items():
    z=load_locked(key,need_features=True); current=z['current']; safe=z['safe']; m5=z['m5']; reason=z['reason']; f=z['features']; strict=reason==2; preserve=reason==1
    paper=np.full(current.shape,np.nan,np.float32); yy=f.y.astype(int).to_numpy(); xx=f.x.astype(int).to_numpy(); paper[yy,xx]=pd.to_numeric(f.paper_value,errors='raise').to_numpy(np.float32)
    residual=np.maximum(paper-current.astype(np.float32),0); hot=np.isfinite(paper)&(residual>=20)
    layers=[(m5&hot,(0,0,255)),(safe&hot,(0,200,0)),(strict&hot,(255,0,0)),(preserve&hot,(0,165,255))]
    image=overlay(current,layers); cv2.imwrite(str(PAGES/key/'retained_residue_overlay.png'),image)
    page_scores.append({'page_key':key,'population':str(r.population),'residual20_pixels':int(hot.sum()),'m5_residual20':int(np.sum(hot&m5)),'safe_residual20':int(np.sum(hot&safe)),'strict_print_residual20':int(np.sum(hot&strict)),'preserve_residual20':int(np.sum(hot&preserve)),'residual_mass':float(residual[hot].sum())})
    del z,current,safe,m5,reason,f,strict,preserve,paper,yy,xx,residual,hot,layers,image
    gc.collect()
page_scores=pd.DataFrame(page_scores).sort_values('residual_mass',ascending=False); page_scores.to_csv(TABLES/'page_residue_ranking.csv',index=False)

def sheet(rows,name):
    rows=rows.head(TOP_PAGES)
    if rows.empty:return
    fig,axes=plt.subplots(len(rows),3,figsize=(15,max(5,5.5*len(rows))),squeeze=False)
    for i,(_,r) in enumerate(rows.iterrows()):
        key=r.page_key
        a=read_gray(SOURCE/'pages'/key/'corrected.png'); b=read_gray(PAGES/key/'historical_m5_assembly.png'); c=cv2.cvtColor(cv2.imread(str(PAGES/key/'retained_residue_overlay.png')),cv2.COLOR_BGR2RGB)
        for j,(im,title) in enumerate([(a,f'{key} current Notebook 72'),(b,f'{key} historical M5 assembly'),(c,f'{key} retained-residue reasons')]):
            axes[i,j].imshow(im,cmap='gray' if im.ndim==2 else None,vmin=0 if im.ndim==2 else None,vmax=255 if im.ndim==2 else None); axes[i,j].set_title(title); axes[i,j].axis('off')
        del a,b,c
    plt.tight_layout(); plt.savefig(VISUAL/f'{name}.png',dpi=125,bbox_inches='tight'); plt.close(fig); gc.collect()
sheet(page_scores,'highest_retained_residue_pages')
sheet(assembly.sort_values('current_vs_historical_different_pixels',ascending=False),'largest_current_vs_historical_assembly_differences')
print(page_scores.head(20).to_string(index=False))
````
